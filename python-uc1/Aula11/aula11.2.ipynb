# Usando fromkeys()

```python
chaves=["nome", "dt_nascimento", "cep", "logradouro", "bairro", "cdd", "estado"]
valor_padrao="N/A"
usuario=dict.fromkeys(chaves, valor_padrao)

print(usuario)
```

```python
usuario["nome"]=input("Digite o seu nome: ")
```

```python
print(usuario)
```

```python
usuario["dt_nascimento"]=input("Digite o sua dt nascimento: ")
```

```python
usuario["cep"]=input("Digite o seu cep: ")
```

```python
usuario["logadouro"]=input("Digite o seu logradouro: ")
usuario["bairro"]=input("Digite seu bairro: ")
usuario["cdd"]=input("Digite sua cdd: ")
usuario["estado"]=input("Digite seu estado: ")

print(usuario)
```

```python
for chaves, valor in usuario.items():
    print(f"{chaves.capitalize():_<20}: {valor.capitalize()}")
```

```python
clientes={}
clientes["12345678912"]=usuario
print(clientes)
```
