## 1- Desafio calculadora basica

```python
num1= int(input("Escolha um numero: "))
num2= int(input("Escolha um segundo numero: "))

opera= int(input("Escolha uma operacao:"))

print("1- para +")
print("2- para -")
print("3- para *")

if opera == 1:
    print(num1 + num2)
elif opera == 2:
    print(num1 - num2)
elif opera == 3:
    print(num1 * num2)
else:
    print("Operacao invalida")
```

# 1.2 Desafio

### Diga se o usuario tem idade para votar

```python
idade= int(input("Digite sua idade: "))

if idade >= 16:
    print("Sua idade é valida a votar")
elif idade <= 16:
    print("Sua idade é invalida a votar")
else:
    print("Idade invalida")
```

# 1.3 Desafio

### Faça um codigo e diga se ele é impar ou par

```python
num= int(input("Digite um numero:"))

if num == 0:
    print("numero é zero")
elif num % 2 == 0:
    print("numero é par")
else:
    print("O numero é impar")
```

# 1.4 Desafio

### Faça um programa que calcule o imc de uma pessoa e exiba sua classificação

```python
peso= float(input("Digite seu peso:"))
alt= float(input("Digite sua altura:"))

imc= (peso/alt**2)

if imc < 18.5:
    print("Vc esta abaixo do peso")
elif imc >= 18.5:
    print("Vc esta no peso ideal")
elif imc <= 29.9:
    print("Vc esta em sobrepeso")
else:
    print("Obesidade")
```
