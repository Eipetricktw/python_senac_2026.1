# Exercicio 4

```python
def fatorial(n):
    if n < 0:
        return "Numero invalido para fatorial."
    elif n == 0:
        return 1
    else:
        resultado = 1
    for i in range(1, n+1):
        resultado *= i
        return resultado

print("fatorial de 5 é:", fatorial(5))
```

# Exercicio 7

```python
def eh_polindromo(palavra):
    return palavra == palavra[::-1]

print("Ana é palindromo?", eh_polindromo("ana"))
print("Python é palindromo?", eh_polindromo("python"))
```

# Exercicio 8

```python
def contar_vogais(texto):
    vogais = "aeiouAEIOU"
    contador = 0
    for char in texto:
        if char in vogais:
            contador += 1
    return contador
# Exemplo de chamada
print("Número de vogais em 'Python':", contar_vogais("Python"))
```
