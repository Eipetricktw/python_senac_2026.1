## 8.1 Funções simples

```python
def soma (a, b):
    soma=a+b
    return soma
```

```python
resultado=soma(12, 15)
print(f"{resultado:4.2f}")
```

```python
def soma_2 (n1, n2):
    soma=n1+n2
    print(soma)
```

```python
soma_2(12.4, 15)
```

```python
resposta=soma_2(12, 50)
print(resposta)
```

```python
def potencia (numero, expoente):
    resultado=numero**expoente
    return resultado
```

```python
numero=2
expoente=8

print(potencia(numero, expoente))
```

# Funções 2

```python
def saudacao(display="sem nome"):
    print(f"Ola, {display}")
```

```python
saudacao("Pet")
```

```python
saudacao()
```

# Passagem de argumentos arbitrario

```python
def saudacoes(*nomes):
    #print(nome)
    for nome in nomes:
        saudacao(nome)
```

```python
saudacoes("Luis", "Rodrigo", "Papagaio")
```
