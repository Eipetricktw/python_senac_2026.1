# Desafio 1

```python
produtos = (
    ("Arroz", 5.99),
    ("Feijão", 7.49),
    ("Leite", 4.89),
    ("Óleo", 9.99),
    ("Açúcar", 3.29)
)

print("LISTA DE PRODUTOS")
for nome, preco in produtos:
    print(f"{nome:.<20} R$ {preco:.2f}")
```

### Desafio 1.2

```python
produtos = (
    ("Arroz", 5.99, 2),
    ("Feijão", 7.49, 4),
    ("Leite", 4.89, 2),
    ("Óleo", 9.99, 2),
    ("Açúcar", 3.29, 5)
)

valor_produto=0
valor_total=0

print("LISTA DE PRODUTOS")
for nome, preco, qtd in produtos:
    valor_produto= preco*qtd
    valor_total=valor_total+valor_produto
    print(f"{nome:.<20} R$ {preco:.2f}, Qtd: {qtd}, R$:{valor_produto:8.2f}")

print(f"\n\tValor total da compra: R$ {valor_total:7.2f}")
```

# Desafio 3

```python
numeros = (2, 7, 10, 13, 17, 22, 29, 31, 40, 53)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primos = tuple(num for num in numeros if eh_primo(num))
print(f"Números primos: {primos}")
```

### Desafio 3.2

```python
numeros = (2, 7, 10, 13, 17, 22, 29, 31, 40, 53)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primos_lista=[]
for num in numeros:
    if eh_primo(num):
        primos_lista.append(num)

print(f"Numeros primos: {primos_lista}")
primos=tuple(primos_lista)
print(f"Números primos: {primos}")
```

# Desafio 9

```python
import random

# Gerando os 5 lançamentos com for tradicional
lancamentos_lista = []
for _ in range(5):
    lancamentos_lista.append(random.randint(1, 6))

lancamentos = tuple(lancamentos_lista)
print(f"Resultados: {lancamentos}")

# Contagem tradicional
for num in range(1, 7):
    print(f"{num} saiu {lancamentos.count(num)} vezes.")
```
