# Desafio 11

```python
vetor = [1, 2, 2, 3, 4, 4, 5]

vetor_sem_duplicatas = list(dict.fromkeys(vetor))

print("Vetor sem duplicatas:", vetor_sem_duplicatas)
```

```python
vetor = [1, 2, 2, 3, 4, 4, 5]

vetor_sem_duplicatas=[]

for numero in vetor:
    if numero not in vetor_sem_duplicatas:
        vetor_sem_duplicatas.append(numero)

print("Vetor sem duplicatas:", vetor_sem_duplicatas)
```
