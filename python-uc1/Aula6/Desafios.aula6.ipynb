```python
print(f"Os numeros pares de 1 a 20 são:")
for num in range(1, 21):
    if num % 2 == 0:
        print(f"{num:2d}", end=",")
```

# Desafio 3.2

```python
print(f"Os numeros impares de 1 a 20 são:")
for num in range(1, 21):
    if num % 2 == 1:
        print(f"{num:2d}", end=",")
```

# Desafio 3.3

```python
pares=[]
impares=[]

for numero in range(1, 21):
    if (numero % 2==0):
        pares.append(numero)
    else:
        impares.append(numero)
print(f"Numeros pares:   {pares}")
print(f"Numeros impares: {impares} ")
```
