# Desafio 4.2

```python
import random
segredo=random.randint(0, 10)
tentativas=0
while tentativas < 3:
    numero=int(input("Digite um numero: "))
    if numero == segredo:
        print("Parabens! vc acertou!")
        break
    else:
        print("Vc errou, tente novamente")
    tentativas=tentativas+1
print(f"O numero secreto era {segredo}")
```
