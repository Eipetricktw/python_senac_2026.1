#  Persistência de Dados em Arquivos Simples

A persistência de dados em arquivos simples é uma das formas mais diretas e acessíveis de salvar informações em Python, sem a complexidade de configurar um sistema de gerenciamento de banco de dados completo.
Ela é amplamente utilizada no desenvolvimento de sistemas para arquivos de configuração, logs de sistema, exportação de relatórios e troca rápida de dados.
Abaixo, detalho os principais formatos utilizados no ecossistema Python, acompanhados de suas vantagens, limitações e exemplos práticos comentados.

#### Formato TXT (Texto Puro)
O formato TXT armazena dados em formato de texto bruto, sem estrutura rígida pré-definida. A organização do conteúdo depende inteiramente da lógica do seu código.

Vantagens: Extrema simplicidade, compatibilidade universal e leveza.

Limitações: Ausência de estrutura nativa; exige que você faça manualmente a lógica de parsing (leitura e interpretação) linha por linha.

Quando utilizar: Em arquivos de log de eventos, anotações rápidas ou arquivos de texto livre.

#### Formato CSV (Comma-Separated Values)
O formato CSV armazena dados tabulares (linhas e colunas), onde os valores são separados por um caractere delimitador (geralmente vírgula , ou ponto e vírgula ;).

Vantagens: Ideal para planilhas; suporte nativo em Python através do módulo csv; amplamente aceito pelo Excel, Google Sheets e ferramentas de BI.

Limitações: Suporta apenas dados estruturados bidimensionais (tabelas simples); não lida bem com dados aninhados (estruturas complexas dentro de outras).

Quando utilizar: Exportação de relatórios financeiros, listagens de usuários ou importação de cadastros em massa para sistemas Django.

#### Formato TXT: Filtragem de Logs
Ideal para textos livres e registros de eventos. A leitura é feita linha por linha.

```python
# Simulando a leitura de um arquivo de log para extrair erros
erros_encontrados = []

# 'r' abre o arquivo para leitura com codificação segura UTF-8
with open("sistema.log", "r", encoding="utf-8") as arquivo_txt:
    for linha in arquivo_txt:
        # Filtramos apenas as linhas que contêm a palavra 'ERRO'
        if "ERRO" in linha:
            # .strip() remove espaços em branco e quebras de linha extras
            erros_encontrados.append(linha.strip())

print("Erros filtrados do TXT:", erros_encontrados)
```

#### Formato CSV: Dados Tabulares e DictReader

Ideal para planilhas e dados estruturados em linhas e colunas. O DictReader transforma cada linha diretamente em um dicionário Python.

```python
import csv

# Suponha um arquivo 'usuarios.csv' com as colunas: nome, email, ativo
with open("usuarios.csv", "r", encoding="utf-8") as arquivo_csv:
    # Transforma o CSV em uma lista de dicionários usando o cabeçalho como chave
    leitor = csv.DictReader(arquivo_csv)
    
    for linha in leitor:
        # Acessamos os dados de forma limpa usando as chaves do cabeçalho
        print(f"Usuário: {linha['nome']} | E-mail: {linha['email']}")
```

#### Formato JSON e Conversão de CSV para JSON
O JSON mapeia perfeitamente para dicionários e listas do Python, sendo ideal para APIs e configurações. Vamos ver um exemplo completo de conversão de CSV para JSON:

import csv
import json

# 1. Lemos os dados de um arquivo CSV
with open("produtos.csv", "r", encoding="utf-8") as arquivo_csv:
    produtos_lista = list(csv.DictReader(arquivo_csv))

# 2. Salvamos esses mesmos dados em formato JSON estruturado
# indent=4 deixa o arquivo legível para humanos; ensure_ascii=False preserva acentos
with open("produtos.json", "w", encoding="utf-8") as arquivo_json:
    json.dump(produtos_lista, arquivo_json, indent=4, ensure_ascii=False)

print("Conversão de CSV para JSON realizada com sucesso!")

### Atualizando Dados em arquivos JSON (O mais prático)
Como o JSON se transforma diretamente em dicionários e listas do Python, ele é o formato mais fácil de atualizar. Você carrega os dados, altera a chave desejada e salva de volta.

```python
import json

# Suponha que temos um arquivo 'config.json' com as configurações do sistema:
# {"modo_manutencao": false, "tema": "claro"}

# 1. LEITURA: Carregamos o arquivo JSON para um dicionário Python
try:
    with open("config.json", "r", encoding="utf-8") as arquivo:
        dados = json.load(arquivo)
except FileNotFoundError:
    dados = {}

# 2. MODIFICAÇÃO: Atualizamos o valor da chave 'tema' na memória
dados["tema"] = "escuro"
dados["modo_manutencao"] = True

# 3. ESCRITA: Reescrevemos o arquivo inteiro com os dados atualizados
with open("config.json", "w", encoding="utf-8") as arquivo:
    json.dump(dados, arquivo, indent=4, ensure_ascii=False)

print("Arquivo JSON atualizado com sucesso!")
print("Novos dados:", dados)
```

#### Atualizando Dados em arquivos CSV (Manipulando Tabelas)
Para atualizar um arquivo CSV (como uma lista de produtos ou usuários), precisamos seguir o ciclo padrão: ler todas as linhas, encontrar a linha específica que desejamos alterar, modificar os dados na memória e reescrever o arquivo completo.

Imagine que temos um arquivo produtos.csv e queremos atualizar o preço de um produto específico.

```python
import csv

produto_alvo = "Teclado Mecânico"
novo_preco = "199.90"
linhas_atualizadas = []
encontrado = False

# 1. LEITURA e MODIFICAÇÃO na memória
with open("produtos.csv", "r", encoding="utf-8") as arquivo_csv:
    leitor = csv.reader(arquivo_csv)
    
    # Capturamos o cabeçalho para não perdê-lo
    cabecalho = next(leitor)
    linhas_atualizadas.append(cabecalho)
    
    for linha in leitor:
        # A linha é uma lista, ex: ['Teclado Mecânico', '150.00', '10']
        # Verificamos se o nome do produto (coluna 0) é o que procuramos
        if linha[0] == produto_alvo:
            linha[1] = novo_preco  # Atualizamos o preço (coluna 1)
            encontrado = True
            
        # Adicionamos a linha (modificada ou original) à nossa nova lista
        linhas_atualizadas.append(linha)

# 2. REESCRITA: Salvamos a lista completa de volta no arquivo CSV
if encontrado:
    with open("produtos.csv", "w", newline="", encoding="utf-8") as arquivo_csv:
        escritor = csv.writer(arquivo_csv)
        escritor.writerows(linhas_atualizadas)
    print(f"O preço de '{produto_alvo}' foi atualizado com sucesso!")
else:
    print(f"Produto '{produto_alvo}' não encontrado no arquivo.")
```

#### Conclusão
Arquivos simples (TXT, CSV e JSON) são excelentes soluções leves para configurações, logs e troca rápida de dados, mas possuem limitações severas de concorrência e escalabilidade.
