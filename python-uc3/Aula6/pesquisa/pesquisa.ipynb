## Lista 

#### O que é: É uma coleção ordenada e mutável (modificável) de itens. É a estrutura de dados mais versátil em Python?

Ordenada: Os itens mantêm a ordem em que foram inseridos e podem ser acessados por índices numéricos (começando em 0).
Mutável: Você pode adicionar, remover ou alterar elementos após a lista ter sido criada.
Permite duplicatas: Como os elementos são indexados, listas podem conter itens com o mesmo valor.
Sintaxe: São definidas usando colchetes

#### Quando usar?
Usamos listas sempre que precisamos de:

Flexibilidade: Adicionar, remover ou alterar elementos (.append(), .remove()).
Ordem: Manter a sequência exata em que os itens foram inseridos (filas, rankings, históricos).
Acesso por índice: Buscar ou modificar itens usando a posição numérica (lista[0]).

```python
# Exemplo básico: uma lista contendo diferentes itens de compras
lista_compras = ["arroz", "feijão", "leite", "café"]
print(lista_compras)
```

# Tupla

#### O que é?

Uma tupla é uma coleção ordenada, porém imutável (ou seja, você não pode alterar, adicionar ou remover elementos após criá-la).

Sintaxe: É definida usando parênteses () (ou até mesmo sem eles, apenas separando por vírgulas).
Exemplo: coordenadas = (-22.5085, -43.1729)

#### Quando usar?

Imutabilidade: Quando você tem um conjunto de dados fixos que o programa não deve modificar de forma alguma por engano, ex: dias da semana.
Desempenho: Como as tuplas são imutáveis, o Python otimiza o uso de memória e o processamento delas, tornando-as um pouco mais rápidas que as listas.
Chaves de Dicionário: Como são imutáveis, tuplas podem ser usadas como chaves em dicionários (o que não é permitido para listas).

# Dicionário

#### O que é?
Um dicionário é uma coleção mutável e não ordenada que armazena dados em pares de chave e valor, key: value.

Sintaxe: É definido usando chaves {} separando a chave do valor por dois-pontos (:).

```python
usuario = {
    "nome": "Ana",
    "idade": 30,
    "email": "ana@email.com"
}
```

### Quando usar no desenvolvimento de sistemas?

Representação de entidades: É o formato padrão para representar registros do mundo real onde cada campo tem um nome descritivo em vez de um número de índice.
Lookups: Se você precisa procurar informações rapidamente usando um identificador único, os dicionários fazem isso de forma extremamente eficiente.
Manipulação de JSON: Como o formato JSON da internet se traduz perfeitamente em dicionários no Python, eles são essenciais para APIs e desenvolvimento web.

# Set

#### O que é?
Um set é uma coleção não ordenada, mutável, mas que não permite elementos duplicados.

Sintaxe: É definido usando chaves {} (sem os dois-pontos de chave-valor dos dicionários) ou a função set().

```python
emails_unicos = {"ana@email.com", "carlos@email.com", "ana@email.com"}
```

#### Quando usar no desenvolvimento de sistemas?

Remover Duplicatas: Se você tem uma lista cheia de valores repetidos e precisa extrair apenas os valores únicos rapidamente, basta converter a lista para um set.
Testes de Pertencimento Rápidos: Saber se um item existe em um set é muito mais rápido do que em uma lista, especialmente com grandes volumes de dados.
Operações Matemáticas de Conjuntos: Ideal para sistemas que precisam cruzar dados, ver quais itens faltam, etc.

# Persistência de Dados

Persistência de Dados (Salvando em Arquivos e Bancos)
Arquivos de Texto e CSV (open(), módulo csv): Para salvar planilhas simples ou logs de texto.

Arquivos JSON (json): Padrão da web para salvar configurações e dados estruturados em formato de texto.

Bancos de Dados Relacionais (SQLite via módulo sqlite3 ou ORMs como SQLAlchemy): Permite criar tabelas, fazer consultas complexas (SQL) e manter os dados salvos de forma segura em um arquivo local ou servidor.

Bancos de Dados NoSQL / Chave-Valor (ex: TinyDB, Pickle): Para salvar objetos Python diretamente em arquivos de forma rápida.

# Serialização de Objetos (JSON e Pickle)

### Arquivos JSON (json)
#### O que é?

O JSON (JavaScript Object Notation) é um formato universal de texto leve baseado em pares de chave-valor e listas. Embora tenha nascido no ecossistema web, o Python possui um módulo nativo chamado json que lê e escreve esses arquivos com facilidade.

O JSON traduz estruturas do Python para uma string de texto formatada que pode ser salva em um arquivo

```python
import json

# Dados em memória (Dicionário/Lista)
usuario = {
    "nome": "Carlos",
    "idade": 28,
    "ativo": True,
    "habilidades": ["Python", "SQL"]
}

# Salvando (persistindo) em um arquivo
with open("usuario.json", "w", encoding="utf-8") as arquivo:
    json.dump(usuario, arquivo, ensure_ascii=False, indent=4)

# Lendo (recuperando) do arquivo após reiniciar o programa
with open("usuario.json", "r", encoding="utf-8") as arquivo:
    dados_carregados = json.load(arquivo)

print(dados_carregados["nome"])  # Saída: Carlos
```

#### Quando usar no desenvolvimento de sistemas?

Arquivos de Configuração: Salvar preferências de usuários ou configurações de um sistema.
Comunicação com APIs: Como a maioria absoluta das APIs web envia e recebe dados em JSON, saber manipulá-lo é obrigatório para desenvolvedores Python.
Interoperabilidade: O JSON é legível por humanos e pode ser aberto por qualquer outra linguagem.

### Módulo Pickle (pickle)
#### O que é?

O Pickle é um módulo nativo do Python usado para serialização de objetos em formato binário. Diferente do JSON, o Pickle compacta quase qualquer objeto do Python incluindo instâncias de classes personalizadas, funções e estruturas complexas — e joga tudo em um arquivo binário.

```python
import pickle

class Carro:
    def __init__(self, modelo):
        self.modelo = modelo

# Criando um objeto complexo
meu_carro = Carro("Fusca")

# Salvando (persistindo) em um arquivo binário (.pkl)
with open("carro.pkl", "wb") as arquivo:
    pickle.dump(meu_carro, arquivo)

# Lendo (recuperando) o objeto com todos os seus atributos intactos
with open("carro.pkl", "rb") as arquivo:
    carro_recuperado = pickle.load(arquivo)

print(carro_recuperado.modelo)  # Saída: Fusca
```

# Bancos de Dados (SQL)

Quando combinamos listas em Python com bancos de dados SQLite, o cenário mais comum é pegar dados de uma lista e salvá-los no banco, consultar o banco de dados e receber os registros em formato de lista

import sqlite3

# 1. Conectando ao banco de dados (será criado um arquivo 'projeto.db' na pasta)
conexao = sqlite3.connect("projeto.db")
cursor = conexao.cursor()

# 2. Criando a tabela de tarefas se ela ainda não existir
cursor.execute("""
    CREATE TABLE IF NOT EXISTS tarefas (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        descricao TEXT NOT NULL
    )
""")

# 3. Nossa lista de dados (cada item é uma tupla contendo os valores para a linha)
lista_novas_tarefas = [
    ("Configurar o ambiente virtual",),
    ("Criar as models no Django",),
    ("Escrever testes automatizados",)
]

# 4. Inserindo todos os itens da lista de uma vez no banco
cursor.executemany("INSERT INTO tarefas (descricao) VALUES (?)", lista_novas_tarefas)

# Salvamos (commit) as alterações no banco de dados
conexao.commit()
print("Dados da lista inseridos no SQLite com sucesso!")

No Django, você raramente precisa escrever comandos SQL crus como acima, pois o ORM do Django faz isso por você de forma orientada a objetos.
